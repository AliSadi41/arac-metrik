```python
import pandas as pd

# Bir üst klasöre çıkıp (../) data/raw içine bakıyoruz
df_tez = pd.read_csv("../data/raw/arac_verisi.csv")

print(df_tez.shape)
df_tez.head()
```

```python
# Sütunların tiplerine ve eksik (null) veri olup olmadığına bakalım
print("--- VERİ TİPLERİ VE EKSİK VERİ DURUMU ---")
df_tez.info()

print("\n--- SAYISAL VERİLERİN İSTATİSTİKSEL ÖZETİ ---")
# Sayısal verilerin (fiyat, km, yıl) özetini alalım. 
# round(2) ile virgülden sonraki küsuratları sadeleştiriyoruz ki okuması kolay olsun.
df_tez.describe().round(2)
```

```python
# Veri setimizde en çok ilanı olan ilk 10 markayı (brand) listeleyelim
df_tez['brand'].value_counts().head(10)
```

```python
# 2011 model Hyundai Getz 1.5 dizel araçları filtreliyoruz
getz_2011 = df_tez[(df_tez['brand'] == 'hyundai') & 
                   (df_tez['model'] == 'getz') & 
                   (df_tez['year'] == 2011) & 
                   (df_tez['variant'].str.contains('1.5', case=False, na=False))]

print(f"Pazardaki 2011 Getz 1.5 sayısı: {len(getz_2011)}")

# Sonuçları fiyata göre sıralayıp en uygun fiyatlı ilk 5 aracı görelim
getz_2011[['variant', 'km', 'price', 'city']].sort_values(by='price').head()
```
